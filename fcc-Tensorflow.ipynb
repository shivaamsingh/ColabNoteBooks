```python
rank1_tensor=tf.Variable(["Test","ok","Kim","moew"],tf.string)
rank2_tensor=tf.Variable([["test","ok","moew"],["test","yes","woof"],["ok","my","bad"]],tf.string)
```

```python
print(tf.rank(number))
print(tf.rank(rank1_tensor))
print(tf.rank(rank2_tensor))
```

#changing shape

```python
tensor1= tf.ones([1,2,3]) #tf.ones() creates a shape [1,2,3] tensor full of ones

tensor2= tf.reshape(tensor1,[2,3,1])#reshapes existing data to shape [2,3,1]
tensor3= tf.reshape(tensor2,[3,-1]) #-1 tells the tensor to calculate the size of the dimension in that place
                                    #this will reshape the tensor to [3,2]

#the no. of elements in the reshaped tensor must match the number in the original
```

```python
print(tensor1)
print(tensor2)
print(tensor3)
#notice the changes in shape
```
